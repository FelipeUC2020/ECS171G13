### Demo
This is a demo notebook that loads a model from a checkpoint or pickle file through the `load_model` function and uses it to make predictions on a test set. A call is going to be made to the LLM to get recommendations based on the model's predictions.

```python
from data_cleanup import DataProcessor
from model_loader import ModelLoader
from llm_recommendations import get_recommendations
```

```python
# load data
INPUT_STEPS = 24*3
OUTPUT_STEPS = 24

processor = DataProcessor(input_steps=INPUT_STEPS, output_steps=OUTPUT_STEPS)
(X_train, y_train), (X_val, y_val), (X_test, y_test) = processor.load_and_process_data()
```

```python
# get the 4 best channels from the data: 
channel_indices = [4,5,6,7]
n_features = X_train.shape[2]
assert max(channel_indices) < n_features, 'Channel indices out of range for X_train features'

X_train_ch = X_train[:, :, channel_indices]  # shape (n_samples, time, 4)
X_val_ch   = X_val[:, :, channel_indices]
X_test_ch  = X_test[:, :, channel_indices]

print('Selected channel shapes:')
print('X_train_ch', X_train_ch.shape, 'y_train', y_train.shape)
```

```python
# initialize model
model = ModelLoader("CNN")

predictions = model.get_predictions(X_test_ch[0])
print(predictions)

get_recommendations(predictions)
```
